## Re-Labelling the types
Relabel the types so there's only 4 categories : Call, Song, Drumming, other


```python
def split_types(x):
    """Split a 'type' cell into normalized values."""
    if not isinstance(x, str):
        return []
    return [t.strip(' .').lower() for t in x.split(',') if t.strip(' .')]
```

```python
all_types = metadata_df['type'].apply(split_types).explode().dropna()
unique_types = sorted(all_types.unique())

print(all_types.value_counts())  # same normalization as the mapping

categories = {'c': 'Call', 's': 'song', 'd': 'drumming', 'o': 'other'}
mapping = {}

for i, value in enumerate(unique_types, 1):
    while True:
        choice = input(f"[{i}/{len(unique_types)}] '{value}' -> (c)all (s)ong (d)rumming (o)ther: ").strip().lower()
        if choice in categories:
            mapping[value] = categories[choice]
            break
        print("Invalid choice, try again.")
```

```python
import json; json.dump(mapping, open(os.path.join(DATA_DIR, 'type_mapping.json'), 'w'), indent=2)
```

```python
metadata_df['type_clean'] = metadata_df['type'].apply(
    lambda x: sorted({mapping[t] for t in split_types(x)})
)
```
